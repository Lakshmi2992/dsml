```python
#EXP-4
#a) Load the iris.csv dataset using Pandas and display the first five records of the dataset.
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Iris.csv")
print("First five records:")
print(df.head())
```

```python
#b) Create a scatter plot showing the relationship between sepal.length and sepal.width.
colors = {
    "Iris-setosa": "red",
    "Iris-versicolor": "green",
    "Iris-virginica": "blue"
}



plt.figure(figsize=(7, 5))

plt.scatter(
    df["SepalLengthCm"],
    df["SepalWidthCm"],
    color="purple"
)

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Sepal Length vs Sepal Width")
plt.grid(True)
plt.show()
```

```python
#c) Create a scatter plot of the Iris dataset using different colors for each flower variety (Setosa,Versicolor, and Virginica).
plt.figure(figsize=(7, 5))

for species, color in colors.items():
    data = df[df["Species"] == species]
    plt.scatter(
        data["SepalLengthCm"],
        data["SepalWidthCm"],
        color=color,
        label=species
    )
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Sepal Length vs Sepal Width by Species")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#d) Create a line plot showing the relationship between sepal.length and sepal.width.
plt.figure(figsize=(8, 5))

plt.plot(
    df["SepalLengthCm"],
    df["SepalWidthCm"],
    marker="o",
    linestyle="-",
    color="orange"
)

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Line Plot of Sepal Length vs Sepal Width")
plt.grid(True)
plt.show()
```

```python
#e) Create a histogram showing the distribution of sepal.length values.
plt.figure(figsize=(7, 5))

plt.hist(
    df["SepalLengthCm"],
    bins=10,
    color="skyblue",
    edgecolor="black"
)

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.title("Distribution of Sepal Length")
plt.grid()
plt.show()
```

```python
#f) Create a bar chart showing the total sepal.length for each Iris flower variety.
total_sepal_length = df.groupby("Species")["SepalLengthCm"].sum()

plt.figure(figsize=(7, 5))

plt.bar(
    total_sepal_length.index,
    total_sepal_length.values,
    color=["red", "green", "blue"]
)

plt.xlabel("Species")
plt.ylabel("Total Sepal Length (cm)")
plt.title("Total Sepal Length by Iris Species")
plt.xticks(rotation=15)
plt.show()
```

```python
#g) Create a box plot to visualize the distribution of sepal.length for different Iris flower varieties.
plt.figure(figsize=(7, 5))

data_box = [
    df[df["Species"] == "Iris-setosa"]["SepalLengthCm"],
    df[df["Species"] == "Iris-versicolor"]["SepalLengthCm"],
    df[df["Species"] == "Iris-virginica"]["SepalLengthCm"]
]

plt.boxplot(
    data_box,
    labels=["Setosa", "Versicolor", "Virginica"]
)

plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
plt.title("Distribution of Sepal Length by Species")
plt.grid(axis="y")
plt.show()
```

```python
#h) Create a pie chart showing the percentage distribution of the three Iris flower varieties in the dataset.
species_counts = df["Species"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    species_counts.values,
    labels=species_counts.index,
    autopct="%1.1f%%",
    colors=["red", "green", "blue"],
    startangle=90
)

plt.title("Percentage Distribution of Iris Species")
plt.show()
```

```python
#i) Create a scatter plot comparing petal.length and petal.width, using different colors for each Iris variety.
import matplotlib.pyplot as plt

colors = {
    "Iris-setosa": "red",
    "Iris-versicolor": "green",
    "Iris-virginica": "blue"
}

plt.figure(figsize=(7, 5))

for species, color in colors.items():
    data = df[df["Species"] == species]

    plt.scatter(
        data["PetalLengthCm"],
        data["PetalWidthCm"],
        color=color,
        label=species
    )

plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("Petal Length vs Petal Width by Species")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#j) Create multiple subplots displaying the following histograms:
#Histogram of sepal.length
#Histogram of sepal.width
#Histogram of petal.length
#Histogram of petal.width


fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Histogram of Sepal Length
axes[0, 0].hist(
    df["SepalLengthCm"],
    bins=10,
    color="red",
    edgecolor="black"
)
axes[0, 0].set_title("Histogram of Sepal Length")
axes[0, 0].set_xlabel("Sepal Length (cm)")
axes[0, 0].set_ylabel("Frequency")


# Histogram of Sepal Width
axes[0, 1].hist(
    df["SepalWidthCm"],
    bins=10,
    color="green",
    edgecolor="black"
)
axes[0, 1].set_title("Histogram of Sepal Width")
axes[0, 1].set_xlabel("Sepal Width (cm)")
axes[0, 1].set_ylabel("Frequency")


# Histogram of Petal Length
axes[1, 0].hist(
    df["PetalLengthCm"],
    bins=10,
    color="blue",
    edgecolor="black"
)
axes[1, 0].set_title("Histogram of Petal Length")
axes[1, 0].set_xlabel("Petal Length (cm)")
axes[1, 0].set_ylabel("Frequency")


# Histogram of Petal Width
axes[1, 1].hist(
    df["PetalWidthCm"],
    bins=10,
    color="orange",
    edgecolor="black"
)
axes[1, 1].set_title("Histogram of Petal Width")
axes[1, 1].set_xlabel("Petal Width (cm)")
axes[1, 1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
#k) Create a grouped bar chart comparing the average sepal.length and sepal.width for each Iris flower variety.
average_sepal = df.groupby("Species")[
    ["SepalLengthCm", "SepalWidthCm"]
].mean()

x = range(len(average_sepal))
width = 0.35

plt.figure(figsize=(8, 5))

plt.bar(
    [i - width / 2 for i in x],
    average_sepal["SepalLengthCm"],
    width=width,
    label="Average Sepal Length",
    color="skyblue"
)

plt.bar(
    [i + width / 2 for i in x],
    average_sepal["SepalWidthCm"],
    width=width,
    label="Average Sepal Width",
    color="orange"
)

plt.xticks(
    list(x),
    ["Setosa", "Versicolor", "Virginica"]
)

plt.xlabel("Species")
plt.ylabel("Average Length/Width (cm)")
plt.title("Average Sepal Length and Sepal Width by Species")
plt.legend()
plt.show()
```

```python
#l) Create a bubble scatter plot using petal.length and petal.width, where the size of each bubble represents sepal.length.

plt.figure(figsize=(8, 6))

for species, color in colors.items():
    data = df[df["Species"] == species]

    plt.scatter(
        data["PetalLengthCm"],
        data["PetalWidthCm"],
        s=data["SepalLengthCm"] * 30,
        color=color,
        alpha=0.6,
        label=species
    )

plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("Bubble Plot: Petal Length vs Petal Width")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#exp-6
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

#1. Load the dataset using Pandas.
dataset = pd.read_csv("/content/Iris.csv")
print(dataset.head())

#2. Separate the features and target variable.
x = dataset.iloc[:, :-1].values
y = dataset.iloc[:, 5].values

#3. Split the dataset into 80% training and 20% testing data.
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=0)

#4. Apply StandardScaler for feature scaling.
scalar = StandardScaler()
scalar.fit(x_train)
x_train = scalar.transform(x_train)
x_test = scalar.transform(x_test)

#5. Train a k-NN classifier with k = 5.
classifier = KNeighborsClassifier(n_neighbors=5)
classifier.fit(x_train, y_train)

#6. Predict the classes of the test data.
y_pred = classifier.predict(x_test)

print("Actual values:")
print(y_test)

print("\nPredicted values:")
print(y_pred)

#7. Display the confusion matrix, classification report, and accuracy.
print("\nConfussion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100,"%")
```

```python
#exp-7
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Load the dataset using Pandas
dataset = pd.read_csv("/content/Cancer_Data - Cancer_Data.csv")
print(dataset.head())

# 2. Separate the features and target variable(this method also can be used)

# Remove id and diagnosis from the dataset
x = dataset.drop(['id', 'diagnosis'], axis=1)

# Diagnosis is the target variable
y = dataset['diagnosis']

# 3. Split the dataset into 80% training and 20% testing data
x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.20,
    random_state=0
)

# 4. Apply StandardScaler for feature scaling
scalar = StandardScaler()

scalar.fit(x_train)

x_train = scalar.transform(x_train)
x_test = scalar.transform(x_test)

# 5. Train a k-NN classifier with k = 5
classifier = KNeighborsClassifier(n_neighbors=5)

classifier.fit(x_train, y_train)

# 6. Predict the classes of the test data
y_pred = classifier.predict(x_test)

print("Actual values:")
print(y_test)

print("\nPredicted values:")
print(y_pred)

# 7. Display confusion matrix,
# classification report and accuracy

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100, "%")
```

```python
#pg8
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score


# Read the dataset
data = pd.read_csv("/content/WineQT.csv")

# A) Exploratory Data Analysis


# Display first 5 records
print("First 5 Records:")
print(data.head())


# Check missing values
print("\nMissing Values:")
print(data.isnull().sum())


# Display descriptive statistics
print("\nDescriptive Statistics:")
print(data.describe())


# Display wine quality distribution
print("\nWine Quality Distribution:")
print(data["quality"].value_counts())

#scatterplot
plt.scatter(data["alcohol"], data["pH"])
plt.xlabel("Alcohol")
plt.ylabel("pH")
plt.title("Alcohol vs pH")
plt.show()


# Histogram of Alcohol
plt.hist(data["alcohol"], bins=20)
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.title("Alcohol Distribution")
plt.show()


# Box plot of Alcohol grouped by Quality
data.boxplot(column="alcohol", by="quality")
plt.xlabel("Quality")
plt.ylabel("Alcohol")
plt.title("Alcohol by Wine Quality")
plt.suptitle("")
plt.show()
# C) k-NN Classification
# Remove missing values
data = data.dropna()
# Separate features and target
X = data.drop("quality", axis=1)
y = data["quality"]
# Split data into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=0
)
# Standardize the data
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Different k values
print("\nAccuracy for different k values:")

for k in [1, 3, 5, 7, 9]:

    model = KNeighborsClassifier(n_neighbors=k)

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    print("k =", k, "Accuracy =", accuracy * 100, "%")



# User enters k value


k = int(input("\nEnter k value: "))

model = KNeighborsClassifier(n_neighbors=k)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy for k =", k, ":", accuracy * 100, "%")
```

```python
#EXP->9(1)
#Using the Breast_Cancer.csv dataset:
#1. Perform basic EDA by displaying the first five records, missing values,descriptive statistics, and diagnosis distribution.

import pandas as pd
df = pd.read_csv("/content/Cancer_Data - Cancer_Data.csv")

# First five records
print("First Five Records:")
print(df.head())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Descriptive statistics
print("\nDescriptive Statistics:")
print(df.describe())

# Diagnosis distribution
print("\nDiagnosis Distribution:")
print(df["diagnosis"].value_counts())

#2. Plot a scatter plot of radius_mean vs texture_mean, a histogram of radius_mean, and a box plot of radius_mean by diagnosis.

import matplotlib.pyplot as plt
import seaborn as sns
sns.scatterplot(
    data=df,
    x="radius_mean",
    y="texture_mean",
    hue="diagnosis"
)

plt.title("Radius Mean vs Texture Mean")
plt.xlabel("Radius Mean")
plt.ylabel("Texture Mean")
plt.show()

plt.hist(
    df["radius_mean"],
    bins=30,
    color="skyblue",
    edgecolor="black"
)

plt.title("Histogram of Radius Mean")
plt.xlabel("Radius Mean")
plt.ylabel("Frequency")
plt.show()

sns.boxplot(
    data=df,
    x="diagnosis",
    y="radius_mean"
)

plt.title("Radius Mean by Diagnosis")
plt.xlabel("Diagnosis")
plt.ylabel("Radius Mean")
plt.show()


#3. Apply Gaussian Naïve Bayes to predict diagnosis using the numerical features,with an 80:20 train-test split.

from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB


X = df.drop(
    columns=["id", "diagnosis", "Unnamed: 32"],
    errors="ignore"
)

y = df["diagnosis"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Create Gaussian Naive Bayes classifier
classifier = GaussianNB()

classifier.fit(X_train, y_train)

y_pred = classifier.predict(X_test)

print("Predicted values:")
print(y_pred)

#4. Display the classification report and accuracy.

from sklearn.metrics import classification_report, accuracy_score

# Classification report
print("Classification Report:")
print(classification_report(y_test, y_pred))

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100, "%")
```

```python
#EXP->9(2)
#Using the Iris.csv dataset:
#1.Perform basic EDA by displaying the first five records, missing values,descriptive statistics, and species distribution.
import pandas as pd

df = pd.read_csv("/content/Iris.csv")

print("First Five Records:")
print(df.head())
print("\nMissing Values:")
print(df.isnull().sum())
print("\nDescriptive Statistics:")
print(df.describe())
print("\nSpecies Distribution:")
print(df["Species"].value_counts())

#2. Plot a scatter plot of Sepal Length vs Sepal Width, a histogram of SepalLength, and a box plot of Sepal Length by Species.
import matplotlib.pyplot as plt
import seaborn as sns
sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="SepalWidthCm",
    hue="Species"
)
plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()

plt.hist(
    df["SepalLengthCm"],
    bins=20,
    color="skyblue",
    edgecolor="black"
)
plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()

sns.boxplot(
    data=df,
    x="Species",
    y="SepalLengthCm"
)

plt.title("Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length")
plt.xticks(rotation=15)
plt.show()


#3. Apply Gaussian Naïve Bayes to predict Species using the numerical features,with an 80:20 train-test split.
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB

X = df[
    [
        "SepalLengthCm",
        "SepalWidthCm",
        "PetalLengthCm",
        "PetalWidthCm"
    ]
]
y = df["Species"]
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
classifier = GaussianNB()
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
print("Predicted values:")
print(y_pred)

#4.Display the classification report and accuracy

from sklearn.metrics import classification_report, accuracy_score
print("Classification Report:")
print(classification_report(y_test, y_pred))
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100, "%")


```

```python
#EXP->9(3)
#Wine Quality Analysis and Naïve Bayes Classification
#Using the Wine_Quality.csv dataset:
#1. Perform basic EDA by displaying the first five records, missing values,descriptive statistics, and quality distribution.

import pandas as pd
df = pd.read_csv("/content/WineQT.csv")
print("First Five Records:")
print(df.head())
print("\nMissing Values:")
print(df.isnull().sum())
print("\nDescriptive Statistics:")
print(df.describe())
print("\nQuality Distribution:")
print(df["quality"].value_counts())

#2. Plot a scatter plot of alcohol vs pH, a histogram of alcohol, and a box plot of alcohol by quality.

import matplotlib.pyplot as plt
import seaborn as sns
sns.scatterplot(
    data=df,
    x="alcohol",
    y="pH",
    hue="quality"
)
plt.title("Alcohol Vs pH")
plt.xlabel("Alcohol")
plt.ylabel("pH")
plt.show()

plt.hist(
    df["alcohol"],
    bins=20,
    color="skyblue",
    edgecolor="black"
)
plt.title("Histogram of Alcohol")
plt.xlabel("alcohol")
plt.ylabel("Frequency")
plt.show()

sns.boxplot(
    data=df,
    x="quality",
    y="alcohol"
)

plt.title("Alcohol by Wine Quality")
plt.xlabel("Quality")
plt.ylabel("Alcohol")
plt.xticks(rotation=15)
plt.show()

#3. Apply Gaussian Naïve Bayes to predict quality using the numerical features,with an 80:20 train-test split.

from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB

X = df[
    [
        "fixed acidity",
        "volatile acidity",
        "citric acid",
        "residual sugar",
        "chlorides",
        "free sulfur dioxide",
        "total sulfur dioxide",
        "density",
        "pH",
        "sulphates",
        "alcohol"
    ]
]
y = df["quality"]
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
classifier = GaussianNB()
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
print("Predicted values:")
print(y_pred)

#4. Display the classification report and accuracy.

from sklearn.metrics import classification_report, accuracy_score
print("Classification Report:")
print(classification_report(y_test, y_pred))
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100, "%")
```
